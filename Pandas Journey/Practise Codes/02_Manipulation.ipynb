# Data Manipulation by selecting rows and cols

## Selecting Columns and Rows

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohan"],
    "Age": [25, 28, 24, 30, 26],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 50000, 40000, 60000, 48000]
}

df = pd.DataFrame(data)
```

## Select a single/multiple column

```python
# Selecting a single columns
df["Name"]
# Selecting multiple columns
df[["Name","Age","Salary"]]
```

## Creating a new Column

```python
df["Bonus"] = df['Salary'] * .1
print(df)
```

### Create a column using a condition
Suppose employees earning at least ₹50,000 are classified as "High" earners and others as "Standard".

```python
df["Salary_Level"] = "Standard" # created with default and next we will change the row values not cols
print(df)
```

```python
df.loc[df["Salary"] >= 50000, "Salary_Level"] = "High"
print(df)
```

## Modify an existing column
Suppose the company gives everyone a salary increase of ₹2,000.

```python
df['Salary'] = df['Salary'] + 2000
print(df)
```

## Removing a column 

```python
df.drop('Age',axis=1)
```

Here you can see that Age is not completely removed in the below cell.

```python
print(df)
```

To remove completely, we use inplace = True

```python
# df.drop('Level',axis=1,inplace=True)
# succesfully dropped the column 'Level' from the dataframe but modern way is :
df.drop(columns=['Level'], inplace=True)
```

```python
print(df)
# Level has been removed from the dataframe.
```

## Remove a row

```python
df.drop(3,axis = 0)
```

```python
print(df)
```

## Select a row using loc and iloc

loc selects rows using their index labels.
loc basically is location.

```python
print(df)
```

```python
df.loc[1]
```

### Select multiple rows

```python
df.loc[1:3]
```

### You can select a particular row and column together

```python
df.loc[2, "Name"]
```

```python
df.loc[0:3, ["Name", "Salary"]]
```

### Iloc - same as loc but for loc[2:4] both are inclusive but for iloc the ending position is excluded.

```python
df.iloc[2:4]
```

Only accepts the integer position

```python
df.iloc[0:3, 0:2]
```
