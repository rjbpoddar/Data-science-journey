# Filtering data with conditions

## Conditional Selection

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohan"],
    "Age": [25, 28, 24, 30, 26],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 50000, 40000, 60000, 48000]
}

df = pd.DataFrame(data)
```

### Find employees whose salary is greater than ₹45,000.

```python
df[df['Salary']> 45000]
```

### Find employees whose salary is greater than ₹45,000 and age is greater than 28

```python
df[(df['Salary']>45000) & (df["Age"]>28)]
```

### Find employees who work in HR AND are older than 26.


```python
df[(df['Department']=="HR") & (df["Age"]>26)]
```

### Find employees whose salary is less than 45000 OR whose age is greater than 28.


```python
df[(df['Salary']<45000) | (df["Age"]>28)]
```

### Find employees who do NOT work in IT.


```python
df[(df['Department']!="IT")]
```

### Find employees who work in IT AND have a salary greater than or equal to 40000.

```python
df[(df['Department']=="IT") & (df["Salary"]>=40000)]
```
