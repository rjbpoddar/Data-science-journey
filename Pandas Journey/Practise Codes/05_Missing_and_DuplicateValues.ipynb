# Working with null & Duplicate values

## Create the dataframe

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohan"],
    "Age": [25, None, 24, 30, None],
    "Salary": [45000, 50000, None, 60000, 48000]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

## Find Missing values

```python
df.isna()
# This will return a boolean DataFrame indicating which values are missing (True) or not (False).
```

```python
df.isna().sum()
# This will return how many null values are there in each column
```

## Fill missing values

```python
df.fillna(2)
# All the null values will be replaced with the args value eg, 2
```

### Suppose you want to replace missing ages with 0

```python
df["Age"] = df["Age"].fillna(0)
print(df)
```

### To replace missing salaries with the average salary

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print(df)
```

### Remove rows containing missing values using dropna()
By default, this returns only rows that have no missing values in any column.

```python
df.dropna()
```

# Working with duplicate values

## Creating duplicate dataframe

```python
# creating a dataset with duplicates
import pandas as pd
data = {
    "Name": ["Rahul", "Priya", "Rahul", "Amit"],
    "Department": ["IT", "HR", "IT", "IT"],
    "Salary": [45000, 50000, 45000, 40000]
}

df = pd.DataFrame(data)
print(df)
```

## Identify Duplicates 
This returns a Boolean Series. By default, the first occurrence is marked False, and subsequent matching rows are marked True.

```python
df.duplicated()
```

```python
print(df.duplicated().sum())
```

## Remove Duplicates
Important: By default, drop_duplicates() compares all columns. If two rows have the same name but different salaries, they are not complete duplicates.

```python
df.drop_duplicates()
```

```python
df
```

## Check duplicates based on selected columns

```python
df
```

```python
df.duplicated(subset=['Name','Department'])
```
