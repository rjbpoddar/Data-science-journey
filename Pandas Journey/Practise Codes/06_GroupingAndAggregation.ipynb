# Grouping and aggregating data with groupby()

Imagine your manager asks:
- What is the average salary in each department?
- How many employees work in each department?
- What is the highest salary in each department?
***Instead of manually filtering each department, we can use groupby()***

## Creating a dataframe

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 50000, 40000, 60000, 48000],
    "Age": [25, 28, 24, 30, 26]
}

df = pd.DataFrame(data)
print(df)
```

## Calculate average salary by department

```python
df.groupby("Department")["Salary"].mean()
# the one which we want to use group in first bracket and aggregate fun in 3rd bracket
```

## Minimum, maximum and average salary in one operation

```python
df.groupby("Department")["Salary"].agg(["min", "max", "mean"])
```

## value_counts() and sorting results
how to count occurrences of values and sort the results — two common operations in data analysis.

By default, **value_counts() sorts the counts from highest to lowest.**

### Count values using value_counts()


```python
df["Department"].value_counts()
```

### Calculate percentages instead of counts

```python
df['Department'].value_counts(normalize=True)*100
```

### Sort a groupby() result

```python
df.groupby("Department")["Salary"].mean().sort_values(ascending=False)
```
