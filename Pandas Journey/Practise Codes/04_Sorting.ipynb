# Sorting & Filtering

## Creating dataframe

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohan"],
    "Age": [25, 28, 24, 30, 26],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 50000, 40000, 60000, 48000]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

## Sorting 

### Sort employees by salary

```python
df.sort_values(by="Salary")
```

### Sort employees by salary DESC or highest salary first

```python
df.sort_values(by="Salary", ascending=False)
```

## Filtering using isin()
It's cleaner than writing multiple equality conditions with | when checking several possible values.

### Find employees from either the IT or HR department

```python
df[df['Department'].isin(['HR','IT'])]
```

### Select employees whose department is NOT in IT or HR using isin() and ~.

```python
df[~(df["Department"].isin(["IT", "HR"]))]
```
