# Pandas Overview 
Pandas is a python library that is used to manipulate, clean and analyze strcutured data.
Pandas can hold anytype of datatype.

## The two most important Pandas structures
    1. Series
    2. DataFrame

# Series
Series in Pandas can be compared with a single column of a spreadsheet.
Indexing starts from 0 and we can add manual index names or numbers.


### How to create a Series

```python
import pandas as pd
import numpy as np
```

```python
list = [10,20,30,40]
dict = {
    'name': "Rajib",
    'Age' : 25,
    'Salary' : 28000
}
arr = np.array([20,30,40,50])
```

So, series is a single column. and we can create a series from an array, list and dictionary.

```python
s1 = pd.Series(list)
s2 = pd.Series(dict)
s3 = pd.Series(arr)
print(s1)
print(s2)
print(s3)
```

### How to give custom index names

```python
labels = ['no1','no2','no3','no4']
pd.Series(list, index=labels)
```

# DataFrame
Two-dimensional data.
dataframe is basically the combination of multiple single columns, consists of rows and columns.
A DataFrame is similar to an Excel spreadsheet with rows and columns.

## How to create a DataFrame : Dictionary

```python
data = {
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [25, 28, 24],
    "Marks": [85, 90, 78]
}
df = pd.DataFrame(data)
print(df)
```

Each dictionary key becomes a column name. Each list supplies that column's values.

## How to create a DataFrame : List

```python
list_data = [
    ["Rahul", "Priya", "Amit"],
    [25,28,24],
    [85,90,78]
]
col_heading = ["Name","Age","Marks"]
df2 = pd.DataFrame(list_data, columns=col_heading)
```
