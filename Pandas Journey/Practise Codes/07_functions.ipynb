# Applying functions with apply()

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Rohan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 50000, 40000, 60000, 48000],
    "Age": [25, 28, 24, 30, 26]
}

df = pd.DataFrame(data)
print(df)
```

## Application
### Create a column named Salary_Level: 'High' if salary is at least 50000, otherwise 'Standard'. Use a custom function with apply().

```python
def salary_level(salary):
    if salary >= 50000:
        return "High"
    else:
        return "Standard"

df["Salary_Level"] = df["Salary"].apply(salary_level)
```

## Use apply() and a lambda to create a column named Age_Next_Year, adding 1 to every employee's age.

```python
df['Age_Next_Year'] = df['Age'].apply(lambda x: x+1)
```
